```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import SGDClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Load MNIST
mnist = fetch_openml('mnist_784', version=1, as_frame=False)

X = mnist.data
y = mnist.target.astype(np.uint8)

# Use only 35,000 samples (30k train, 5k test)
X_subset = X[:35000]
y_subset = y[:35000]

X_train, X_test, y_train, y_test = train_test_split(
    X_subset, y_subset, test_size=5000, random_state=42
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)
```

```python
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)
rf_acc = accuracy_score(y_test, rf_pred)

print("Random Forest Accuracy:", rf_acc)
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

knn_pred = knn.predict(X_test)
knn_acc = accuracy_score(y_test, knn_pred)

print("KNN Accuracy:", knn_acc)
```

```python
sgd = make_pipeline(
    StandardScaler(),
    SGDClassifier(random_state=42)
)

sgd.fit(X_train, y_train)
sgd_pred = sgd.predict(X_test)
sgd_acc = accuracy_score(y_test, sgd_pred)

print("SGD Accuracy:", sgd_acc)
```

```python
nb = GaussianNB()
nb.fit(X_train, y_train)

nb_pred = nb.predict(X_test)
nb_acc = accuracy_score(y_test, nb_pred)

print("Naive Bayes Accuracy:", nb_acc)
```

```python
voting_clf = VotingClassifier(
    estimators=[
        ('rf', rf),
        ('knn', knn),
        ('sgd', sgd),
        ('nb', nb)
    ],
    voting='hard'
)

voting_clf.fit(X_train, y_train)

voting_pred = voting_clf.predict(X_test)
voting_acc = accuracy_score(y_test, voting_pred)

print("Voting Classifier (Hard) Accuracy:", voting_acc)
```

```python
voting_soft = VotingClassifier(
    estimators=[
        ('rf', rf),
        ('knn', knn),
        ('sgd', sgd),
        ('nb', nb)
    ],
    voting='soft'
)

voting_soft.fit(X_train, y_train)

voting_soft_pred = voting_soft.predict(X_test)
voting_soft_acc = accuracy_score(y_test, voting_soft_pred)

print("Voting Classifier (Soft) Accuracy:", voting_soft_acc)
```

```python
results = {
    "Random Forest": rf_acc,
    "KNN": knn_acc,
    "SGD": sgd_acc,
    "Naive Bayes": nb_acc,
    "Voting (Hard)": voting_acc,
    "Voting (Soft)": voting_soft_acc
}

for model, acc in results.items():
    print(f"{model}: {acc:.4f}")
```

Through this activity, I learned that ensemble learning combines multiple models to produce a more robust and often more accurate prediction. Instead of relying on a single classifier, the ensemble aggregates predictions, which reduces variance and improves generalization.

I observed that stronger models like Random Forest and KNN already performed very well individually. However, when combined in a voting classifier, the ensemble either slightly improved performance or matched the best model while being more stable overall.
